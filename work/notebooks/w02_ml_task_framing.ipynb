## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Chosen Lane:** Lane 2 — Refresh / Content Opportunity Scoring

**Task Type:** Ranking / Scoring

The purpose is to prioritize which pages to should be reviewed and refreshed first, so the model will rank the pages from highest score to lowest score. This allows SEO team to focus on reviewing highest ranked pages first

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target for this project is priority score which is a proxy derived from search performance such as impressions, clicks, ctr, average position etc. The model predicts this score to rank pages by their priority for review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@K is the main success metric. It measures the proportion of the top K pages that are actually high priority pages. Higher Precision@K score indicates that the model performs well at prioritizing pages that need review.
